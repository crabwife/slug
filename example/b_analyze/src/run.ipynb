```python
import csv
from pathlib import Path
from statistics import mean
```

```python
TASK = Path.cwd().parent
```

```python
with (TASK/'input/snails.csv').open(newline='') as stream:
    leaves = [int(row['leaves']) for row in csv.DictReader(stream)]
```

```python
(TASK/'output').mkdir(exist_ok=True)
with (TASK/'output/summary.csv').open('w', newline='') as stream:
    writer = csv.writer(stream, lineterminator='\n')
    writer.writerow(['snail_count', 'mean_leaves'])
    writer.writerow([len(leaves), mean(leaves)])
```
